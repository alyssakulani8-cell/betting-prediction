```python
# Generate synthetic data for illustration
np.random.seed(42)
n = 5000
synth = pd.DataFrame({
    'home_win_rate': np.random.uniform(0.2, 0.8, n),
    'away_win_rate': np.random.uniform(0.2, 0.8, n),
    'elo_diff': np.random.randn(n) * 100 + 50,
    'home_form_decay': np.random.uniform(0, 1, n),
    'home_rest_days': np.random.exponential(5, n),
    'target': np.random.choice(['H', 'D', 'A'], n, p=[0.46, 0.24, 0.30]),
})
synth.head()
```

```python
fig, ax = plt.subplots(2, 2, figsize=(14, 10))

ax[0,0].hist(synth['home_win_rate'], bins=30, alpha=0.7, color='#3b82f6')
ax[0,0].set_title('Home Win Rate Distribution')

ax[0,1].hist(synth['elo_diff'], bins=30, alpha=0.7, color='#22c55e')
ax[0,1].set_title('ELO Difference Distribution')

for target, color, label in [('H', '#3b82f6', 'Home'), ('D', '#f59e0b', 'Draw'), ('A', '#ef4444', 'Away')]:
    subset = synth[synth['target'] == target]
    ax[1,0].hist(subset['home_form_decay'], bins=20, alpha=0.5, color=color, label=label)
ax[1,0].set_title('Form Decay by Outcome')
ax[1,0].legend()

ax[1,1].hist(synth['home_rest_days'], bins=30, alpha=0.7, color='#a855f7')
ax[1,1].set_title('Rest Days Distribution')

plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap
corr = synth.select_dtypes(include=[np.number]).corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdBu_r', center=0)
plt.title('Feature Correlation Matrix')
plt.show()
```

```python
synth['target'].value_counts().plot(kind='bar', color=['#3b82f6', '#f59e0b', '#ef4444'])
plt.title('Target Distribution (Home / Draw / Away)')
plt.xticks(rotation=0)
plt.show()
```
